# Data Analysis Example

This notebook demonstrates basic data analysis techniques using pandas and numpy.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("Libraries imported successfully!")
```

## 1. Generate Sample Data

```python
# Generate sample dataset
np.random.seed(42)

n_samples = 1000
data = {
    'date': pd.date_range(start='2023-01-01', periods=n_samples, freq='D'),
    'sales': np.random.normal(100, 20, n_samples).clip(0),
    'visitors': np.random.poisson(50, n_samples),
    'category': np.random.choice(['A', 'B', 'C'], n_samples),
    'region': np.random.choice(['North', 'South', 'East', 'West'], n_samples)
}

df = pd.DataFrame(data)
df['conversion_rate'] = (df['sales'] / df['visitors']).clip(0, 1)

print(f"Dataset shape: {df.shape}")
df.head()
```

## 2. Basic Statistics

```python
# Summary statistics
df.describe()
```

```python
# Group by category
category_summary = df.groupby('category').agg({
    'sales': ['mean', 'median', 'std'],
    'visitors': ['mean', 'sum'],
    'conversion_rate': 'mean'
}).round(2)

category_summary
```

## 3. Visualizations

```python
# Time series plot
fig, ax = plt.subplots()
df.set_index('date')['sales'].rolling(window=7).mean().plot(ax=ax, label='7-day MA')
ax.set_title('Sales Over Time (7-day Moving Average)')
ax.set_xlabel('Date')
ax.set_ylabel('Sales')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Distribution plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales distribution
axes[0].hist(df['sales'], bins=30, edgecolor='black', alpha=0.7)
axes[0].set_title('Sales Distribution')
axes[0].set_xlabel('Sales')
axes[0].set_ylabel('Frequency')

# Box plot by category
df.boxplot(column='sales', by='category', ax=axes[1])
axes[1].set_title('Sales by Category')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Sales')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```
